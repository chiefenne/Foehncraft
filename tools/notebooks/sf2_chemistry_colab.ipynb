# Foehncraft – chemistry of the flow cells on a Colab GPU

Runs the GPU check of the steam furnace stage SF2b: the reactors of the flow cells (the RODAS3 integration of one time step
per cell) as a single-source Rust kernel on the GPU of this session. A row of 4096 cells of stoichiometric methane and air
(900 to 2400 K, fresh, half and fully burnt) is integrated once on the CPU threads and once on the GPU, for GRI-Mech 3.0 and
for Jones & Lindstedt, and the end states and step counts are compared **bitwise**. The report also gives the time per cell
on both.

1. **Runtime → Change runtime type → T4 GPU** (any GPU of compute capability 7.0 or newer works).
2. Set `BRANCH` below if the chemistry branch is not merged yet (`furnace-chemistry-flow-sf2b`).
3. **Runtime → Run all.** Takes about 10 minutes, most of it compiling.
4. Copy the report printed at the end into the thread.
